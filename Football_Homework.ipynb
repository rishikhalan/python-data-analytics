```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
```

```python
data = {
    "Player": ["Messi", "Ronaldo", "Neymar", "Mbappe",
               "Haaland", "Salah", "Vinicius", "Kane"],

    "Age": [39, 41, 34, 27, 26, 34, 26, 33],

    "Country": ["Argentina", "Portugal", "Brazil", "France",
                "Norway", "Egypt", "Brazil", "England"],

    "Goals": [850, 960, 450, 350, 300, 280, 220, 320],

    "Club": ["Inter Miami", "Al Nassr", "Santos", "Real Madrid",
             "Man City", "Liverpool", "Real Madrid", "Bayern Munich"]
}

df = pd.DataFrame(data, index = [1, 2, 3, 4, 5, 6, 7, 8])
df
```

```python
top_8 = df.sort_values("Goals", ascending = False)
top_8
```

```python
plt.figure(figsize = (10, 6), facecolor = "lightyellow")
plt.bar(top_8["Player"], top_8["Goals"], color = [
    "#1E3A5F",  # Navy Blue
    "#6EC6E8",  # Sky Blue
    "#7BAE7F",  # Sage Green
    "#F2C94C",  # Golden Yellow
    "#F2994A",  # Warm Orange
    "#EB5757",  # Coral Red
    "#BB9CE8",  # Lavender
    "#FFF4D6"   # Cream
], hatch = ["/", "*", "x", ".", "/", "*", "x", "."], edgecolor = "#000000", label = "Goals", zorder = 2)
plt.yticks(range(0, 1001, 100))
plt.title("Football's Top 8", fontsize = 20)
plt.legend()
plt.ylabel("Goals", fontsize = 15)
plt.xlabel("Players", fontsize = 15)
plt.grid(alpha = 0.3, linestyle = "--")
plt.show()
```
