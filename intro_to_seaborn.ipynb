```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = {
    "Name": [
        "Aarav", "Liam", "Noah", "Ethan", "Lucas", "Arjun",
        "Oliver", "Ryan", "Aiden", "Kabir", "Daniel", "Rohan",
        "Leo", "Adam", "Ishaan", "Maya", "Sophia", "Emma",
        "Ava", "Anaya", "Mia", "Zara", "Chloe", "Ella",
        "Aria", "Diya"
    ],

    "Age": [
        24, 31, 28, 35, 29, 42, 26, 33, 38, 27, 45, 30,
        25, 39, 32, 23, 34, 29, 41, 26, 36, 28, 31, 37,
        24, 43
    ],

    "Gender": [
        "Male", "Male", "Male", "Male", "Male", "Male",
        "Male", "Male", "Male", "Male", "Male", "Male",
        "Male", "Male", "Male", "Female", "Female", "Female",
        "Female", "Female", "Female", "Female", "Female", "Female",
        "Female", "Female"
    ],

    "Salary": [
        42000, 58000, 51000, 72000, 61000, 85000, 47000, 68000,
        79000, 53000, 92000, 60000, 45000, 81000, 67000, 44000,
        69000, 57000, 88000, 49000, 75000, 62000, 54000, 73000,
        46000, 83000
    ],

    "Department": [
        "IT", "Finance", "Marketing", "HR", "IT", "Finance",
        "Sales", "IT", "Management", "Marketing", "Finance", "Sales",
        "IT", "HR", "Sales", "Marketing", "IT", "Finance",
        "Management", "HR", "Sales", "Marketing", "IT", "Finance",
        "HR", "Management"
    ]
}

df = pd.DataFrame(data)
df
```

```python
avgsalary = df.groupby("Department")["Salary"].mean()
avgsalary
```

```python
plt.figure(figsize = (8,5))
plt.bar(avgsalary.index, avgsalary.values)
plt.title("average salary by department")
```

```python
sns.barplot(x = "Department", y = "Salary", data = df, errorbar = None, palette = "icefire")
plt.title("average salary by department")

```
