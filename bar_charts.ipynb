```python
import matplotlib.pyplot as plt # run this before anything
import pandas as pd # run this before anything!!!
import numpy as np # run this before anything!!!

days = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]
hours = [1.75, 1.75, 1.75, 1.75, 0.75, 1.75, 2.25]
colors = ["red", "orange", "yellow", "green", "blue", "indigo", "violet"]
```

```python
plt.figure(figsize = (10, 6))
plt.bar(days, hours, color = colors, edgecolor = 'black', linewidth = 3)
plt.xlabel("Days of the Week", fontsize = 14)
plt.ylabel("Hours on a screen", fontsize = 14)
plt.title("My Avg Weekly Screen Time", fontsize = 20)
```

```python
regions = ["North", "East", "West", "South"]
sales = [57000, 65600, 78600, 41800]
plt.figure(figsize = (10, 6))
plt.bar(regions, sales, color = ["red", "orange", "yellow", "green"], edgecolor = 'black', linewidth = 3)
plt.xlabel("Regions of India", fontsize = 14)
plt.ylabel("Revenue", fontsize = 14)
plt.title("Distribution of phones in India over 5 yrs(fake data)", fontsize = 20)
```

```python
products = ["Product A", "Product B", "Product C"]
north_sales = [200, 150, 180]
south_sales = [250, 200, 230]
east_sales = [300, 250, 280]
west_sales = [350, 300, 330]

bar_width = 0.15
index = np.arange(len(products))

plt.figure(figsize = (10, 6))
plt.bar(index, north_sales, bar_width, color = "blue", label = "North")
plt.bar(index+bar_width, south_sales, bar_width, color = "gold", label = "South")
plt.bar(index+2*bar_width, east_sales, bar_width, color = "green", label = "East")
plt.bar(index+3*bar_width, west_sales, bar_width, color = "red", label = "West")
plt.legend()
plt.xticks(index+bar_width, products)
plt.xlabel("Products", fontsize = 14)
plt.ylabel("Sales in Units", fontsize = 14)
plt.title("Distribution of Products in India over 3 months(fake data)", fontsize = 20)
```

```python
# Data for revenue loss/profit in different months
months = ["Jan", "Feb", "Mar", "Apr", "May"]
profits_and_losses = [1000, -500, 200, -800, 300]

plt.figure(figsize = (10, 6))
plt.bar(months, profits_and_losses, color = ["lightgreen" if x >= 0 else "red" for x in profits_and_losses], edgecolor = 'black', linewidth = 3)
plt.xlabel("Months", fontsize = 14)
plt.ylabel("Profit and Losses", fontsize = 14)
plt.title("Data for revenue loss/profit in different months(fake data)", fontsize = 20)
```

```python
# Data for population growth in different countries

countries = [
    "USA",
    "India",
    "China",
    "Germany",
    "Brazil",
    "Japan",
    "Canada",
    "Australia",
    "France",
    "Mexico",
    "UK",
    "South Korea"
]

population_growth = [
    2.1,
    1.5,
    0.8,
    0.5,
    1.2,
    0.3,
    1.0,
    1.3,
    0.6,
    1.4,
    0.7,
    0.2
]



plt.figure(figsize = (10, 6), facecolor = "turquoise")
plt.bar(countries, population_growth, color = [
    "red",
    "orange",
    "yellow",
    "green",
    "blue",
    "indigo",
    "violet",
    "pink",
    "brown",
    "black",
    "white",
    "gray"
], edgecolor = 'black', linewidth = 3)
plt.xlabel("Countries", fontsize = 14)
plt.ylabel("Population Growth Percentage", fontsize = 14)
plt.title("Population Growth", fontsize = 20)
plt.xticks(rotation = 30)
plt.grid(True, linestyle = ":", alpha = 0.5)

#inner background
ax = plt.gca()
ax.set_facecolor("lightgrey")
```
