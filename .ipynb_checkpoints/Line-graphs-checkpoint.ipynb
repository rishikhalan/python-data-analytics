```python
import matplotlib.pyplot as plt # run this before anything
import pandas as pd # run this before anything!!!
import numpy as np # run this before anything!!!
```

```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
          "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]
temperatures = [16.8, 19.0, 23.2, 27.1, 29.1, 28.6,
                26.6, 26.0, 25.8, 24.2, 20.6, 17.4]
plt.plot(months, temperatures, marker = "o", linestyle='-', color = "lightgreen")
```

```python
#marker='o': Circle
#marker='s': Square
#marker='d': Diamond
#marker='^': Triangle Up
#marker='v': Triangle Down
#marker='p': Pentagon
#marker='+': Plus
#marker='x': Cross
#marker='*': Star


#Solid Line:

#'-' : Default solid line
#Example: plt.plot(x, y, linestyle='-')
#Dashed Line:

#'--' : A dashed line
#Example: plt.plot(x, y, linestyle='--')
#Dotted Line:

#':' : A dotted line
#Example: plt.plot(x, y, linestyle=':')
#Dash-dot Line:

#'-.' : A combination of dashes and dots
#Example: plt.plot(x, y, linestyle='-.')
```

```python
# Months
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun", 
          "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]

# Average temperatures (°C) for two cities
nyc = [5, 7, 12, 18, 22, 26, 30, 29, 24, 18, 11, 6]   # Example: New York
mumbai = [15, 17, 20, 23, 27, 30, 33, 32, 29, 25, 20, 16] # Example: Mumbai

plt.plot(months, nyc, marker = "s", linestyle='-', color = "red", label = "NYC")
plt.plot(months, mumbai, marker = "o", linestyle='-', color = "blue", label = "MUMBAI")
plt.legend()
plt.grid(True)
plt.xlabel("Months", fontsize = 16)
plt.ylabel("Avg Temperatures (°C)", fontsize = 16)
plt.title("Average Temperatures in Mumbai and NYC", fontsize = 20)
```
