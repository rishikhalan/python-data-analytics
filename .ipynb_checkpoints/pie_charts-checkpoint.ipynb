```python
import matplotlib.pyplot as plt # run this before anything
import pandas as pd # run this before anything!!!
import numpy as np # run this before anything!!!

# Data for smartphone market share
brands = ["Samsung", "Apple", "Xiaomi", "OPPO", "Other"]
market_share = [22, 19, 14, 9, 36]
color = [
    "#6C5CE7",  # Purple
    "#00B894",  # Emerald
    "#0984E3",  # Blue
    "#E84393",  # Pink
    "#FDCB6E"   # Yellow
]

explode = [0.1, 0.1, 0, 0, 0]

plt.figure(figsize = (8, 8))
plt.pie(market_share, labels = brands, autopct = "%1.0f%%", explode = explode, startangle = 160, colors = color, shadow = True)
plt.title("Data for smartphone market share")
```

```python
# Data
toppings = ['Pepperoni', 'Cheese', 'Mushrooms', 'Sausage', 'Olives', 'Others']
votes = [35, 25, 15, 10, 8, 7]

color = [
    "#6C5CE7",  # Purple
    "#00B894",  # Emerald
    "#0984E3",  # Blue
    "#E84393",  # Pink
    "#FDCB6E",   # Yellow
    "red"
]

explode = [0, 0.025, 0, 0, 0, 0]

plt.figure(figsize = (8, 8))
plt.pie(votes, labels = toppings, autopct = "%1.0f%%", explode = explode, startangle = 160, colors = color, wedgeprops = {'linewidth':2,
                                                                                                                         'edgecolor':'white'})
plt.title("What pizza types does Y8 like?")
```
