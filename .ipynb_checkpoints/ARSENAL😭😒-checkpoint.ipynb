```python
df["Match_Score"] = df["Match_Score"].str.replace("-" , "-", regex = False)
```

```python
df["home_goals"] = df["Match_Score"].str[0]
df["away_goals"] = df["Match_Score"].str[2]
df

```

```python
def find_winner(away_goals, home_goals, Home_Team, Away_Team):
    if home_goals > away_goals:
        return Home_Team
    elif home_goals < away_goals:
        return Away_Team
    else:
        return "Draw"

df["Winner"] = ""
for i in range(len(df)):
    df.loc[i, "Winner"] = find_winner(df.loc[i, "away_goals"] , df.loc[i, "home_goals"] , df.loc[i, "Home_Team"], df.loc[i, "Away_Team"])
    
    
```

```python
df["Age_float"] = (
    df["Age"].str.split("-").str[0].astype(float)
    + df["Age"].str.split("-").str[1].astype(float) / 365
)
```
