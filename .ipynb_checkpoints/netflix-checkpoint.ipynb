```python


stuff = df["type"].value_counts()


color = [
    "#6C5CE7",  # Purple
    "#00B894",  # Emerald
]

explode = [0, 0]

plt.figure(figsize = (8, 8))
plt.pie(stuff, labels = ["Movies", "TV Shows"], autopct = "%1.3f%%", explode = explode, startangle = 160, colors = color, shadow = True)
plt.title("Data for Netflix types of shows")
```

```python
ratings = df["rating"].value_counts()

rating = [
    "TV-MA",
    "TV-14",
    "TV-PG",
    "R",
    "PG-13",
    "TV-Y7",
    "TV-Y",
    "PG",
    "TV-G",
    "NR",
    "G",
    "TV-Y7-FV",
    "NC-17",
    "UR",
    "74 min",
    "84 min",
    "66 min"
]


plt.figure(figsize = (10, 6))
plt.bar(ratings, rating, color = "black", edgecolor = 'black', linewidth = 3)
plt.xlabel("Amount of Shows", fontsize = 14)
plt.ylabel("Ratings", fontsize = 14)
plt.title("My Avg Weekly Screen Time", fontsize = 2)
```

```python
import plotly.express as px
```

```python
# Handle multiple countries in one entry
df['country'] = df['country'].fillna('Unknown')
df = df.assign(country=df['country'].str.split(',')).explode('country')
df['country'] = df['country'].str.strip()

# Create the treemap: shows by country and type
fig = px.treemap(df,
                 path=['country', 'type'],
                 title='Treemap of Shows by Country and Type',
                 width=1000,   # increase width
                 height=700)   # increase height

fig.show()
```

```python
# Count number of shows per country
country_counts = df['country'].value_counts().reset_index()
country_counts.columns = ['country', 'count']

# Plot choropleth map
fig = px.choropleth(country_counts,locations='country',locationmode='country names',color='count',title='Number of Shows by Country')
fig.show()
```
