```python
import matplotlib.pyplot as plt # run this before anything
import pandas as pd # run this before anything!!!
import numpy as np # run this before anything!!!
```

```python
movies = ["Hero X", "Shadow Man", "Thunder", "Iron Star", "Night Rider"]
tickets = [450, 620, 380, 710, 560]


colors = ["red", "orange", "yellow", "green", "blue"]

plt.figure(figsize = (10, 6))
plt.bar(movies, tickets, color = colors, edgecolor = 'black', linewidth = 3)
plt.xlabel("Movies", fontsize = 14)
plt.ylabel("Ticket Sales", fontsize = 14)
plt.title("Ticket Sales for Sanketana Cinema", fontsize = 20)
```

```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
subscribers = [1200, 1450, 1800, 2400, 3100, 4200]


plt.plot(months, subscribers, marker = "o", linestyle='-', color = "red")
```

```python
categories = ["Food", "Games", "Books", "Movies", "Savings"]
money = [250, 150, 200, 100, 300]


color = [
    "#6C5CE7",  # Purple
    "#00B894",  # Emerald
    "#0984E3",  # Blue
    "#E84393",  # Pink
    "#FDCB6E"   # Yellow
]

explode = [0, 0, 0, 0, 0]

plt.figure(figsize = (8, 8))
plt.pie(money, labels = categories, autopct = "%1.0f%%", explode = explode, startangle = 160, colors = color, shadow = True)
plt.title('How a Student Spends ₹1000')
```

```python
times = [12, 15, 18, 21, 17, 25, 22, 19, 30, 28,
         16, 14, 24, 20, 26, 31, 18, 23, 27, 21,
         15, 19, 33, 29, 24, 17, 22, 26, 20, 18]

plt.hist(times, bins = 10, edgecolor = "white")
plt.title("Time Taken to Complete a Puzzle")
plt.xlabel("Time in Minutes")
```

```python
countries = ["India", "Japan", "USA", "Brazil", "Australia"]
tourists = [850, 620, 940, 510, 430]


colors = ["red", "orange", "yellow", "green", "blue"]

plt.figure(figsize = (10, 6))
plt.bar(countries, tourists, color = colors, edgecolor = 'black', linewidth = 3)
plt.xlabel("Countries", fontsize = 14)
plt.ylabel("Amount of tourists(thousands)", fontsize = 14)
plt.title("Tourism", fontsize = 20)
```
