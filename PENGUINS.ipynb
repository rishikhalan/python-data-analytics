```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("penguins.csv")
df
```

```python
df.info()
```

```python
sns.relplot(data=df, x = "flipper_length_mm", y = "body_mass_g", kind = "scatter")
```

```python
sns.set_style("darkgrid")
sns.relplot(data=df, x = "flipper_length_mm", y = "body_mass_g", kind = "scatter")
```

```python
sns.set_style("darkgrid")
sns.relplot(data=df, x = "flipper_length_mm", y = "body_mass_g", kind = "scatter", hue = "species")
```

```python
sns.set_style("darkgrid")
sns.relplot(data=df, x = "flipper_length_mm", y = "body_mass_g", kind = "scatter", hue = "species", style = "sex")
```

```python
sns.set_style("darkgrid")
sns.relplot(data=df, x = "flipper_length_mm", y = "body_mass_g", kind = "scatter", hue = "sex", col = "species")
```

```python
sns.barplot(data = df, x  = "species", y = "body_mass_g")
```

```python
sns.catplot(data = df, x = "species", y = "body_mass_g", hue = "sex", kind = "bar")
```

```python
sns.heatmap(df.corr(numeric_only = True), annot = True, cmap = "PiYG")
```

```python
sns.pairplot(df, hue = "species")
plt.suptitle("Pairplot of penguin measurements")
```
