```python
data1 = {"Cricketer" : ["Sachin Tendulkar", "Virat Kohli", "Mahendra S Dhoni"],
       "Total Runs" : [36691, 37695, 22705]
       }
d1 = pd.DataFrame(data1, index = [1, 2, 3])
d1
```

```python
d1.plot(kind = "bar", x = "Cricketer", y = "Total Runs", color = ["red", "green", "blue"])
plt.xticks(rotation = 0)
plt.title("INDIAN CRICKETERS TOTAL RUNS IN ALL FORMATS", fontsize = 25)
plt.xlabel("CRICKETERS", fontsize = 15)
plt.ylabel("AMT OF RUNS", fontsize = 15)
plt.show
```
