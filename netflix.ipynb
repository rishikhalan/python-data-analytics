```python
import plotly.express as px
```

```python
# Handle multiple countries in one entry
df['country'] = df['country'].fillna('Unknown')
df = df.assign(country=df['country'].str.split(',')).explode('country')
df['country'] = df['country'].str.strip()

# Create the treemap: shows by country and type
fig = px.treemap(df,
                 path=['country', 'type'],
                 title='Treemap of Shows by Country and Type',
                 width=1000,   # increase width
                 height=700)   # increase height

fig.show()
```

```python
# Count number of shows per country
country_counts = df['country'].value_counts().reset_index()
country_counts.columns = ['country', 'count']

# Plot choropleth map
fig = px.choropleth(country_counts,locations='country',locationmode='country names',color='count',title='Number of Shows by Country')
fig.show()
```
