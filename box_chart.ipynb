```python
import pandas as pd # run this before anything!!!
import matplotlib.pyplot as plt # run this before anything!!!
import numpy as np # run this before anything!!!

# Data for temperatures in different cities
cities = [
    [30, 32, 35, 40, 45, 50, 55, 60, 65, 70, 75, 80],  # City A
    [25, 28, 30, 35, 40, 42, 45, 47, 50, 52, 55, 58],  # City B
    [18, 20, 22, 25, 28, 32, 35, 38, 40, 45, 48, 50],  # City C
    [10, 12, 15, 18, 22, 25, 30, 35, 40, 45, 50, 55],  # City D
]

plt.boxplot(
    cities,
    tick_labels=["City A", "City B", "City C", "City D"],
    patch_artist=True,
    notch=True,
    vert=True,
    widths=0.7
)

plt.title("Temperature Distribution in Different Cities")
plt.xlabel("Cities")
plt.ylabel("Temperature (°C)")

plt.show()
```

```python
# Data for employee salaries in different departments
departments = [
    [55000, 60000, 65000, 70000, 80000, 85000],  # HR
    [75000, 80000, 85000, 90000, 95000],  # Marketing
    [60000, 70000, 75000, 80000, 85000],  # Finance
    [45000, 48000, 52000, 55000, 60000],  # IT
]

plt.boxplot(
    departments,
    tick_labels=["HR", "Marketing", "Finance", "IT"],
    boxprops = dict(facecolor = "orange"),
    patch_artist=True,
    notch=True,
    vert=True,
    widths=0.7
)

plt.title("Employee salaries in different departments")
plt.xlabel("Departments")
plt.ylabel("Employee Salaries")

plt.show()
```

```python

# Data for employee salaries in different departments
departments = [
    [55000, 60000, 65000, 70000, 80000, 85000],  # HR
    [75000, 80000, 85000, 90000, 95000],         # Marketing
    [60000, 70000, 75000, 80000, 85000],         # Finance
    [45000, 48000, 52000, 55000, 60000],         # IT
]

plt.boxplot(
    departments,
    patch_artist=True,
    boxprops=dict(facecolor="lightblue"),
    medianprops=dict(color="red"),
    whiskerprops=dict(color="blue"),
    capprops=dict(color="black")
)

plt.xticks([1, 2, 3, 4], ["HR", "Marketing", "Finance", "IT"])
plt.ylabel("Salary")
plt.title("Employee Salaries by Department")

plt.show()
```
